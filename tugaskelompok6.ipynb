```python
# ============================================================
# PRAKTIKUM PENGOLAHAN CITRA DAN VISI KOMPUTER
# TUGAS PRAKTIKUM KELOMPOK 6
# Garino Wijaya (14)
# Keenan Raditya Prasaja (18)
#
# Menutup:
# 1. MONTH/YEAR
# 2. VALID THRU
# 3. ANGKA VALID THRU
# ============================================================

# Import library
import cv2
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow


# ============================================================
# 1. UPLOAD GAMBAR KTM
# ============================================================

print("Silakan pilih foto KTM...")
uploaded = files.upload()

# Mengambil nama file
filename = list(uploaded.keys())[0]

# Membaca gambar
img = cv2.imread(filename)


# ============================================================
# 2. CEK GAMBAR
# ============================================================

if img is None:
    print("Gambar gagal dibaca!")
else:

    print("Gambar berhasil dibaca!")
    print("Nama file :", filename)
    print("Ukuran gambar :", img.shape[1], "x", img.shape[0], "pixel")


    # ========================================================
    # 3. MENAMPILKAN GAMBAR ASLI
    # ========================================================

    print("\nGambar KTM Asli:")
    cv2_imshow(img)


    # ========================================================
    # 4. MENAMPILKAN GAMBAR DENGAN MATPLOTLIB
    # ========================================================

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 9))
    plt.imshow(img_rgb)
    plt.axis("off")
    plt.title("Gambar KTM Asli")
    plt.show()


    # ========================================================
    # 5. MEMBUAT SALINAN GAMBAR
    # ========================================================

    hasil = img.copy()


    # ========================================================
    # 6. MENUTUP TULISAN MONTH/YEAR
    # ========================================================

    cv2.rectangle(
        hasil,
        (780, 550),      # titik kiri atas
        (1050, 605),     # titik kanan bawah
        (255, 0, 255),   # warna BGR = Ungu
        -1               # -1 = kotak terisi
    )


    # ========================================================
    # 7. MENUTUP TULISAN VALID THRU
    # ========================================================

    cv2.rectangle(
        hasil,
        (680, 600),
        (840, 675),
        (0, 255, 0),     # warna BGR = Hijau
        -1
    )


    # ========================================================
    # 8. MENUTUP ANGKA VALID THRU (02/26)
    # ========================================================

    cv2.rectangle(
        hasil,
        (845, 600),
        (1050, 675),
        (255, 0, 0),     # warna BGR = Biru
        -1
    )


    # ========================================================
    # 9. MENAMPILKAN HASIL DENGAN OPENCV
    # ========================================================

    print("\nHasil KTM Kelompok 6:")
    cv2_imshow(hasil)


    # ========================================================
    # 10. MENAMPILKAN HASIL DENGAN MATPLOTLIB
    # ========================================================

    hasil_rgb = cv2.cvtColor(hasil, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 9))
    plt.imshow(hasil_rgb)
    plt.axis("off")
    plt.title("Hasil Tugas Praktikum Kelompok 6")
    plt.show()


    # ========================================================
    # 11. MENYIMPAN HASIL
    # ========================================================

    output = "KTM_Kelompok_6_Hasil.jpg"
    cv2.imwrite(output, hasil)

    print("\nGambar hasil berhasil disimpan sebagai:")
    print(output)


    # ========================================================
    # 12. DOWNLOAD HASIL
    # ========================================================

    files.download(output)
```
